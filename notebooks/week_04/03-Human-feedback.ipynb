### Import deps

```python
from langsmith import Client
```

### Attatch feedback to specific trace

```python
client = Client()
```

The run ID represents a top level trace ID 

This is the id we will atattch the feedback to

Run IDs contain sub trace IDs for individual pans within the trace

```python
RUN_ID = "019fa330-7f49-7b72-93b5-f05b7c0492f1"
```

Example for simple rating number

```python
client.create_feedback(
    run_id=RUN_ID,
    key="feedback",
    score=0,
    feedback_source_type="api"
) 
```

Example for feedback comment 

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment",
    value="This is a long form free text feedback left by Aurimas.",
    feedback_source_type="api"
)
```

Example comment with score

```python
client.create_feedback(
    run_id=RUN_ID,
    key="comment_with_score",
    value="This is a long form free text feedback left by Aurimas.",
    score=1,
    feedback_source_type="api"
)
```
